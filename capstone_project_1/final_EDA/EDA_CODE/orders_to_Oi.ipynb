```python
import pandas as pd
import matplotlib.pyplot as plt



# 1) merge on order_id (inner join; change to left if you want)
merged = df.merge(oNopNcu, on="order_id", how="inner")

# 2) detect installment column (common candidates)
installment_candidates = ["installments", "payment_installments", "installment_number", "num_installments"]
inst_col = next((c for c in installment_candidates if c in merged.columns), None)
if inst_col is None:
    raise KeyError(f"No installments column found. Checked: {installment_candidates}")

# 3) flags (assume these exact names exist; adjust if different)
never_flag = "flag_never_delivered_x" if "flag_never_delivered_x" in merged.columns else "never_delivered"
very_flag  = "flag_very_slow_x" if "flag_very_slow_x" in merged.columns else ("very_slow_delevery" if "very_slow_delevery" in merged.columns else "very_slow_delivery")

# ensure types

# 4) aggregate per installment
agg = merged.groupby(inst_col).agg(
    total_orders = ("order_id", "nunique"),
    never_count  = (never_flag, "sum"),
    very_count   = (very_flag, "sum")
).reset_index().sort_values(inst_col)

agg["pct_never"] = agg["never_count"] / agg["total_orders"] * 100
agg["pct_very"]  = agg["very_count"]  / agg["total_orders"] * 100

# 5) Plot function: bar (total) + line (%)
def plot_bar_line(x, bar_vals, line_vals, bar_label, line_label, title):
    fig, ax1 = plt.subplots(figsize=(10,5))
    ax1.bar(x, bar_vals, color="C0", alpha=0.7)
    ax1.set_xlabel("Installments")
    ax1.set_ylabel(bar_label, color="C0")
    ax1.tick_params(axis='y', labelcolor="C0")
    ax1.set_xticks(x)
    ax1.set_xticklabels(x, rotation=0)

    ax2 = ax1.twinx()
    ax2.plot(x, line_vals, color="C1", marker="o")
    ax2.set_ylabel(line_label, color="C1")
    ax2.tick_params(axis='y', labelcolor="C1")

    plt.title(title)
    plt.tight_layout()
    plt.show()

# 6) Plot never_delivered
plot_bar_line(
    agg[inst_col].tolist(),
    agg["total_orders"].tolist(),
    agg["pct_never"].tolist(),
    "Total unique orders",
    "% Never Delivered",
    "Never Delivered by Installment (count + %)"
)

# 7) Plot very_slow
plot_bar_line(
    agg[inst_col].tolist(),
    agg["total_orders"].tolist(),
    agg["pct_very"].tolist(),
    "Total unique orders",
    "% Very Slow Delivery",
    "Very Slow Delivery by Installment (count + %)"
)

# 8) Combined percent plot (compare pct_never vs pct_very)
plt.figure(figsize=(10,5))
plt.plot(agg[inst_col], agg["pct_never"], marker="o", label="% Never Delivered")
plt.plot(agg[inst_col], agg["pct_very"], marker="o", label="% Very Slow")
plt.xlabel("Installments")
plt.ylabel("Percent of Orders (%)")
plt.title("Compare: % Never Delivered vs % Very Slow by Installment")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# 9) Quick print of agg table for reference
print(agg.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

df = oNopNcu.copy()

# thresholds
low_val = df["payment_value"].quantile(0.25)
high_val = df["payment_value"].quantile(0.75)

def classify(row):
    # 1) High EMI
    if row["payment_installments"] >= 3:
        return "High EMI shopper"
    
    # 2) COD or Boleto
    if str(row["payment_type"]).lower() in ["cod", "boleto"]:
        return "COD/Boleto shopper"
    
    # 3) Low-value impulse
    if row["payment_value"] <= low_val:
        return "Low-value impulse buyer"
    
    # 4) Premium
    if row["payment_value"] >= high_val:
        return "Premium buyer"
    
    # Otherwise
    return "Regular buyer"

df["segment"] = df.apply(classify, axis=1)

# Pie chart
plt.figure(figsize=(8,8))
df["segment"].value_counts().plot.pie(autopct="%1.1f%%", startangle=90)
plt.title("Customer Segmentation by Payment Behavior")
plt.ylabel("")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = oNopNcu.copy()

# Convert timestamps
cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]
for c in cols:
    df[c] = pd.to_datetime(df[c])

# 1) Seller processing time
df["seller_processing_time"] = (
    df["order_approved_at"] - df["order_purchase_timestamp"]
).dt.days

# 2) Carrier pickup time
df["carrier_pickup_time"] = (
    df["order_delivered_carrier_date"] - df["order_approved_at"]
).dt.days

# 3) Carrier delivery time
df["carrier_delivery_time"] = (
    df["order_delivered_customer_date"] - df["order_delivered_carrier_date"]
).dt.days

# 4) Total delivery completion time
df["total_delivery_time"] = (
    df["seller_processing_time"] +
    df["carrier_pickup_time"] +
    df["carrier_delivery_time"]
)

# 5) State-wise average delivery completion
state_delivery = (
    df.groupby("customer_state")["total_delivery_time"]
    .mean()
    .reset_index()
    .sort_values("total_delivery_time")
)

print(state_delivery)

# 6) Plot: State-wise Avg Delivery Time
plt.figure(figsize=(12,5))
plt.bar(state_delivery["customer_state"], state_delivery["total_delivery_time"], color="purple")
plt.title("State-wise Average Delivery Completion Time (Days)")
plt.xlabel("State")
plt.ylabel("Avg Delivery Time (Days)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
