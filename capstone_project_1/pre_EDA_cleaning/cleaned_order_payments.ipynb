```python
# count before
print("before rows:", len(df), "exact sig-duplicates:",
      df.duplicated(subset=['order_id','payment_value']).sum())

# show some repeated signatures (sample)
print(df[df.duplicated(subset=['order_id','payment_value'], keep=False)].head(10))

# drop duplicates keeping first occurrence (in-place)
df = df.drop_duplicates(subset=['order_id','payment_value'], keep='first').reset_index(drop=True)

# count after
print("after rows:", len(df), "sig-duplicates left:",
      df.duplicated(subset=['order_id','payment_value']).sum())

# (optional) save cleaned payments
df.to_csv('cleaned_order_payments.csv', index=False)
```
