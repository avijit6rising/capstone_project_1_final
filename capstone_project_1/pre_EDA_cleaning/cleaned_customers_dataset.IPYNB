```python
df.rename(columns={"customer_zip_code_prefix" :"customer_zip_code"}, inplace=True)
```

```python
col = "customer_city"                      # <-- yaha apna column naam daal

df[col] = df[col].str.lower().str.strip()  # lowercase + trim spaces
df[col] = df[col].str.replace(r"\s+", " ", regex=True)  # extra spaces hataye

print(df[col].unique())                    # clean values check
df.to_csv("cleaned.csv", index=False)     
```

```python
print("Duplicate rows:", df.duplicated().sum())

# show duplicate rows
dup_rows = df[df.duplicated()]
print(dup_rows.head())

# remove duplicates
df = df.drop_duplicates()
```

```python
df.columns
```

```python
col = "customer_unique_id"

print("Duplicate count:", df.duplicated(subset=[col]).sum())

# rows where column is duplicated
dup_col = df[df.duplicated(subset=[col], keep=False)]
print(dup_col.head())
```

```python
for c in df.columns:
    if df[c].duplicated().sum() > 0:
        print(c, "→", df[c].duplicated().sum(), "duplicates")
```

```python
cols_to_string = ["customer_city", "customer_state"]  
# ↑ yaha apne column naam daal do (jitne chaho)

for c in cols_to_string:
    df[c] = df[c].astype(str).str.strip()
    
display(df)
```
